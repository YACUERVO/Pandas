```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df= pd.read_csv('/Users/Admin/Documents/proyectosAnalisisDatos/Pandas/Data/sales_data.csv')


parts = df['Purchase Address'].astype(str).str.split(',', n=2, expand=True)
df['City'] = parts[1].str.strip()   # parts[1] es la columna 'City'
df.head()
```

```python
producto_conunts = df['Product'].value_counts() 
plt.figure(figsize=(10,6))
plt.pie(producto_conunts, labels=producto_conunts.index, autopct='%1.1f%%', startangle=90)
plt.title('Distribución de productos vendidos')
plt.axis('equal')  # Igualar los ejes para que el círculo sea perfecto
plt.show()
```

```python
# Personalizar colores y agregar sombra
colors = ['skyblue', 'lightgreen', 'salmon', 'gold', 'violet']
explode = [0.1]+ [0] * (len(producto_conunts)-1) # Solo "explotar" el primer segmento

plt.figure(figsize=(4,4))
plt.pie(producto_conunts, labels=producto_conunts.index, autopct='%1.1f%%', startangle=90, colors=colors, explode=explode, shadow=True)
plt.title('Distribución de productos vendidos con personalización')
plt.axis('equal')  # Igualar los ejes para que el círculo sea perfecto
plt.show()
```

```python
#Diagrama de torta con grupos de datos
city_counts = df['City'].value_counts()
plt.figure(figsize=(4,4))
plt.pie(city_counts, labels=city_counts.index, autopct='%1.1f%%', startangle=90)
plt.title('Distribución de ventas por ciudad')
plt.axis('equal')  # Igualar los ejes para que el círculo sea perfecto
plt.show()
```

```python
#Diagrama de torta con porciones pequeñas agrupadas
limite = 8  # Porcentaje mínimo para mostrar individualmente
city_counts['Other'] = city_counts[city_counts < limite].sum()
city_counts = city_counts[city_counts >= limite]
plt.figure(figsize=(4,4))
plt.pie(city_counts, labels=city_counts.index, autopct='%1.1f%%', startangle=90)
plt.title('Distribución de ventas por ciudad (con "Other")')
plt.axis('equal')  # Igualar los ejes para que el círculo sea perfecto
plt.show()
```
