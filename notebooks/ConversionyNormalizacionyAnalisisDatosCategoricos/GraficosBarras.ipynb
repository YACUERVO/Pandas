```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df= pd.read_csv('/Users/Admin/Documents/proyectosAnalisisDatos/Pandas/Data/sales_data.csv')
df.head()
```

```python
plt.figure(figsize=(10,6))
sns.countplot(data=df, x='Product', order=df['Product'].value_counts().index)
plt.title('Frecuencia de produtos vendidos')
plt.xlabel('Producto')
plt.ylabel('Frecuencia')
plt.xticks(rotation=45)
plt.show()
```

```python

plt.figure(figsize=(10,6))
sns.countplot(data=df, y='Product', order=df['Product'].value_counts().index)
plt.title('Frecuencia de produtos vendidos')
plt.xlabel('Cantidad Vendida')
plt.ylabel('Producto')
plt.show()
```

```python
#Barras agrupadas
parts = df['Purchase Address'].astype(str).str.split(',', n=2, expand=True)
df['City'] = parts[1].str.strip()   # parts[1] es la columna 'City'
df
```

```python
#Barras agrupadas
plt.figure(figsize=(10,6))
sns.countplot(data=df, x='City', hue='Product', order=df['City'].value_counts().index)
plt.title('Frecuencia de produtos vendidos por ciudad')
plt.xlabel('Ciudad')
plt.ylabel('Frecuencia')
plt.xticks(rotation=45)
plt.legend(title='Producto', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

```python
# Análisis de proporciones
producto_proporciones = df['Product'].value_counts(normalize=True).reset_index()
producto_proporciones.columns = ['Product', 'Proportion']
display(producto_proporciones)
```

```python
plt.figure(figsize=(10,6))
sns.barplot(data=producto_proporciones, x='Product', y='Proportion', order=producto_proporciones.sort_values('Proportion', ascending=False)['Product'])
plt.title('Proporción de productos vendidos')
plt.xlabel('Producto')
plt.ylabel('Proporción')
plt.xticks(rotation=45)
plt.show()
```

```python
#Personalización avanzada
plt.figure(figsize=(10,6))
sns.countplot(data=df, x='Product', order=df['Product'].value_counts().index, palette='viridis')
plt.title('Frecuencia de produtos vendidos', fontsize=16, color='blue')
plt.xlabel('Producto', fontsize=14)
plt.ylabel('Frecuencia', fontsize=14)
plt.xticks(rotation=45, fontsize=12)
plt.yticks(fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
data = {
    'Category': ['A', 'B', 'C', 'D'],
    'Values': [23, 45, 12, 36]
}
df2 = pd.DataFrame(data)
```

```python
plt.figure(figsize=(4,3)) #Definie el tamaño de la figura
plt.bar(df2['Category'], df2['Values'], color=['red', 'blue', 'green', 'orange']) #Estableciendo colores personalizados
plt.title('Simple Bar Chart') 
plt.xlabel('Category')
plt.ylabel('Values')
plt.grid(axis='y', linestyle='--', alpha=0.7) #Añadiendo una cuadrícula horizontal
plt.show()
```

```python
sns.set_style('whitegrid') #Estableciendo un estilo de fondo
sns.set_palette('pastel') #Estableciendo una paleta de colores  

plt.figure(figsize=(6,4))
sns.barplot(data=df2, x='Category', y='Values')
plt.title('Bar Chart with Seaborn Style')
plt.xlabel('Category')
plt.ylabel('Values')
plt.show()
```

```python
plt.figure(figsize=(6,4))
plt.barh(df2['Category'], df2['Values'], color='skyblue') #Gráfico de barras horizontales
plt.title('Horizontal Bar Chart')
plt.xlabel('Values')
plt.ylabel('Category')
plt.grid(True, axis='x', linestyle='--', alpha=0.7)
plt.show()
```

```python
data_stacked = {
    'Category': ['A', 'B', 'C','D'],
    'Values1': [10, 20, 15,20],
    'Values2': [5, 15, 10,30]
}
df_stacked = pd.DataFrame(data_stacked)
display(df_stacked)
```

```python
plt.figure(figsize=(8,6))
plt.bar(df_stacked['Category'], df_stacked['Values1'], label='Values1', color='lightcoral')
plt.bar(df_stacked['Category'], df_stacked['Values2'], bottom=df_stacked['Values1'], label='Values2', color='lightseagreen')
plt.title('Stacked Bar Chart')  
plt.xlabel('Category')
plt.ylabel('Values')
plt.legend()
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()

    
```

```python
plt.figure(figsize=(6,4))
width = 0.3  

categories = df_stacked['Category']
index = range(len(categories))

plt.bar([i - width/2 for i in index], df_stacked['Values1'], width=width, label='Values1', color='mediumpurple')
plt.bar([i + width/2 for i in index], df_stacked['Values2'], width=width, label='Values2', color='goldenrod')
plt.title('Grouped Bar Chart')
plt.xlabel('Category')
plt.ylabel('Values')
plt.xticks(index, categories) # darle etiquetas correctas al eje x 
plt.legend()
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```
