```python
import pandas as pd
```

```python
df= pd.read_csv('/Users/Admin/Documents/proyectosAnalisisDatos/Pandas/Data/sales_data.csv')
df.head()
```

```python
parts = df['Purchase Address'].astype(str).str.split(',', n=2, expand=True)
df['City'] = parts[1].str.strip()   # parts[1] es la columna 'City'
```

```python
display(df.head())
```

```python
#Frecuencia para productos vendidos
produtos_frecuencia = df['Product'].value_counts()
display(produtos_frecuencia)
```

```python
ciudades_frecuencia = df['City'].value_counts()
display(ciudades_frecuencia)
```

```python
#Proporciones del total de ventas
produtos_proporciones = df['Product'].value_counts(normalize=True)
display(produtos_proporciones)
```

```python
#Proporciones para las ciudades
ciudades_proporciones = df['City'].value_counts(normalize=True)
display(ciudades_proporciones)
```

```python
#Tabla cruzada de frecuencias
tabla_cruzada = pd.crosstab(df['City'], df['Product'])
display(tabla_cruzada)
```

```python
#Agrupar por ciudad y producto
producto_por_ciudad= df.groupby('City')['Product'].value_counts()
display(producto_por_ciudad.head(1000))
```

```python
#Filtro donde los prodcutos sean mayor a 1000 para sacar las proporciones
precio_alto = df[df['Price Each'] > 1000]
precio_alto_proporciones = precio_alto['City'].value_counts(normalize=True)
display(precio_alto_proporciones)
```
