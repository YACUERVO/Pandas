```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = {
    'Meses': ['Enero', 'Febrero', 'Marzo', 'Abril', 'Mayo'],
    'Año': [2015, 2016, 2017, 2018, 2019],
    'Ventas': [230, 150, 500, 350, 200],
    'Beneficios': [20, 30, 50, 70, 90]
}

df = pd.DataFrame(data)
df.head()
```

```python
#Gráfico de líneas simple
plt.figure(figsize=(10,6))
plt.plot(df['Meses'], df['Ventas'], marker='o', label='Ventas')
plt.plot(df['Meses'], df['Beneficios'], marker='s', label='Beneficios')
plt.title('Ventas y Beneficios Mensuales')
plt.xlabel('Meses')
plt.ylabel('Cantidad')
plt.legend()
plt.grid()
plt.show()
```

```python
sns.set(style="whitegrid")
plt.figure(figsize=(10,6))
sns.lineplot(x='Meses', y='Ventas', data=df, marker='o', label='Ventas')
sns.lineplot(x='Meses', y='Beneficios', data=df, marker='s', label='Beneficios')
plt.title('Ventas y Beneficios Mensuales con Seaborn')
plt.xlabel('Meses')
plt.ylabel('Cantidad')
plt.legend()
plt.show()
```

```python
data = {
    'Meses': ['Enero', 'Febrero', 'Marzo', 'Abril', 'Mayo'],
    'Año': [2015, 2016, 2017, 2018, 2019],
    'Producto A': [200, 250, 300, 350, 400],
    'Producto B': [150, 200, 250, 300, 350]
}
df = pd.DataFrame(data)
df.head()
```

```python
plt.figure(figsize=(8,6))
plt.plot(df['Meses'], df['Producto A'], marker='o', label='Producto A', color='blue')
plt.plot(df['Meses'], df['Producto B'], marker='s', label='Producto B', color='orange')
plt.title('Comparación de Ventas: Producto A vs Producto B')
plt.xlabel('Meses')
plt.ylabel('Ventas')
plt.legend()
plt.grid(True, which='both', linestyle='--', linewidth=0.5)
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.lineplot(x='Meses', y='Producto A', data=df, marker='o', label='Producto A', color='blue', errorbar='sd')
sns.lineplot(x='Meses', y='Producto B', data=df, marker='s', label='Producto B', color='orange',errorbar='sd')
plt.title('Comparación de Ventas con Seaborn: Producto A vs Producto B')
plt.xlabel('Meses')
plt.ylabel('Ventas')
plt.legend()
plt.grid(True, which='both', linestyle='--', linewidth=0.5)
plt.show()
```

```python
df_long = pd.melt(df, id_vars=['Meses', 'Año'], value_vars=['Producto A', 'Producto B'], var_name='Producto', value_name='Ventas')  
df_long.head()
```

```python
plt.figure(figsize=(10,6))
sns.lineplot(x='Meses', y='Ventas', hue='Producto', style='Producto', data=df_long, markers=True, dashes=False)
plt.title('Ventas de Productos A y B a lo Largo del Tiempo')
plt.xlabel('Meses')
plt.ylabel('Ventas')
plt.legend(title='Producto')
plt.grid()
plt.show()
```
