```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df= pd.read_csv('/Users/Admin/Documents/proyectosAnalisisDatos/Pandas/Data/sales_data_correlated.csv')


parts = df['Purchase Address'].astype(str).str.split(',', n=2, expand=True)
df['City'] = parts[1].str.strip()   # parts[1] es la columna 'City'
df.head()
```

```python
correlacion_matriz = df[['Order ID','Quantity Ordered','Price Each']].corr()
display(correlacion_matriz)
```

```python
plt.figure(figsize=(5,5))
sns.heatmap(correlacion_matriz, annot=True, cmap='coolwarm', center=0, vmin=-1, vmax=1, linewidths=0.5)
plt.title('Matriz de Correlación')
plt.show()
```

```python
#Analisis de la correlacion
correlaccion_value=df['Price Each'].corr(df['Quantity Ordered'])
print(f'La correlación entre Price Each y Quantity Ordered es: {correlaccion_value}')
```

```python
# Filtrar y mostrar correlaciones fuertes
strong_correlation = correlacion_matriz[(correlacion_matriz > 0.7) | (correlacion_matriz < -0.7)] 
print("Correlaciones fuertes (mayores a 0.7 o menores a -0.7):")
print(strong_correlation)  
```

```python
#Metodos sns para mostrar correlacion 
sns.pairplot(df[['Quantity Ordered','Price Each']])
plt.suptitle('Pairplot de Variables Numéricas', y=1.02)
plt.show()
```
