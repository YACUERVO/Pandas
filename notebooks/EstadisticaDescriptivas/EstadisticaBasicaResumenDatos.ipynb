```python
import pandas as pd
```

```python
df = pd.read_csv('/Users/Admin/Documents/proyectosAnalisisDatos/Pandas/Data/sales_data.csv')
df.head()
```

```python
#Metodo que nos ayuda hacer una descripcion a las columnas numericas
stats = df.describe()
display(stats)
```

```python
#Estadisticas personalizadas 
#Vamos a calcular la media 
mean_price = df['Price Each'].mean()
median_price=df['Price Each'].median()
print(f'Media el precio {mean_price}')
print(f'Mediana el precio {median_price}')
```

```python
#Valores maximos y minimos
min_price=df['Price Each'].min()
max_price=df['Price Each'].max()

print('precio minimo',min_price)
print('precio maximo', max_price)
```

```python
#Distribucion de frecuencias
producto_count = df['Product'].value_counts()
display(producto_count)
```

```python
#Columna categoricas 
df['City']= df['Purchase Address'].apply(lambda x: x.split(',')[1].strip())
display(df)

#Devuelve None (o pd.NA) cuando la fila no es una cadena o no contiene coma.
df['City_1'] = df['Purchase Address'].apply(
    lambda x: x.split(',')[1].strip() if isinstance(x, str) and ',' in x else None
)

# astype(str) evita errores por NaN (aunque convierte nan a 'nan', por eso preferible limpiar antes)

# Es más rápida que .apply porque es vectorizada.
# separa en hasta 3 columnas: Street, City, StateZip
parts = df['Purchase Address'].astype(str).str.split(',', n=2, expand=True)
df['City_2'] = parts[1].str.strip()   # parts[1] es la columna 'City'

# Si las direcciones siguen "calle, ciudad, ESTADO [ZIP]", puedes extraer la ciudad con regex:
df['City_3'] = df['Purchase Address'].str.extract(r'^[^,]+,\s*([^,]+),', expand=False).str.strip()

```

```python
city_counts = df['City'].value_counts()
print(city_counts)
```

```python
#Revision de la distribucion de fechas
df['Order Date']= pd.to_datetime(df['Order Date'], errors='coerce')
date_counts = df['Order Date'].dt.month.value_counts().sort_index()
print(date_counts)
```
