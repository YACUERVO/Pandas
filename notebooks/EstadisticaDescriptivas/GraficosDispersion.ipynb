```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df= pd.read_csv('/Users/Admin/Documents/proyectosAnalisisDatos/Pandas/Data/sales_data_correlated.csv')


parts = df['Purchase Address'].astype(str).str.split(',', n=2, expand=True)
df['City'] = parts[1].str.strip()   # parts[1] es la columna 'City'
```

```python
plt.figure(figsize=(5,3))
plt.scatter(df['Price Each'], df['Quantity Ordered'], alpha=0.5)
plt.title('Gráfico de Dispersión: Quantity Ordered vs Price Each')
plt.xlabel('Price Each')
plt.ylabel('Quantity Ordered')
```

```python
plt.figure(figsize=(5,3))
sns.scatterplot(data=df, x='Price Each', y='Quantity Ordered', alpha=0.5)
plt.title('Gráfico de Dispersión: Quantity Ordered vs Price Each por Ciudad')
plt.xlabel('Price Each')
plt.ylabel('Quantity Ordered')
plt.show()
```

```python
# Variables categoricas de tipo de producto
plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x='Price Each', y='Quantity Ordered', hue='Product' ,alpha=0.5, palette='tab10')
plt.title('Gráfico de Dispersión: Quantity Ordered vs Product por Tipo de Producto')
plt.xlabel('Price Each')
plt.ylabel('Quantity Ordered')

plt.legend(title='Product', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

```python
#Lineas de tendencia
plt.figure(figsize=(8,5))
sns.regplot(data=df,x='Price Each', y='Quantity Ordered',scatter_kws={'alpha':0.5}, line_kws={'color':'red'})
plt.title('Gráfico de Dispersión con Línea de Tendencia: Quantity Ordered vs Price Each')
plt.xlabel('Price Each')
plt.ylabel('Quantity Ordered')
plt.show()
```

```python
#Lineas de tendencia
plt.figure(figsize=(5,3))
sns.scatterplot(data=df,x='Price Each',y='Quantity Ordered',s=40, alpha=0.6, hue='City', palette='Set2')
plt.title('Gráfico de Dispersión con Línea de Tendencia: Quantity Ordered vs Price Each')
plt.xlabel('Price Each')
plt.ylabel('Quantity Ordered')
plt.grid(True, linestyle='--', alpha=0.5 )
plt.show()
```
