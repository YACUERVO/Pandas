```python
import pandas as pd
```

```python
df = pd.DataFrame({
    "fecha_inicio":pd.to_datetime([
        "2025-01-01",
        "2025-01-02",
        "2025-01-03",
        "2025-01-04",
        "2025-01-05",
        "2025-01-06",
        "2025-01-07",
        "2025-01-08",
        "2025-01-09",
        "2025-01-10"
    ]),
    "fecha_final":pd.to_datetime([
        "2025-01-11",
        "2025-01-15",
        "2025-01-01",
        "2025-01-22",
        "2025-01-09",
        "2025-01-02",
        "2025-01-30",
        "2025-01-14",
        "2025-01-15",
        "2025-01-17"
    ])
})

display(df)
```

```python
df['diferencia_dias']=df['fecha_final']- df['fecha_inicio']
display(df)
```

```python
#Restar o sumar periodos para obtener la fecha
df['fecha_7_dia'] = df['fecha_inicio'] + pd.DateOffset(days=7)
df['fecha_1_mes'] = df['fecha_final'] - pd.DateOffset(month=1)
df['fecha_2_años'] = df['fecha_inicio'] + pd.DateOffset(years=2)

display(df)
```

```python
#Filtrar las fechas 
df_filtrado = df[df['fecha_final']>df['fecha_inicio']]
display(df_filtrado)
```

```python
#Oeraciones sobre series temporales
fecha = pd.date_range(start='2024-01-01', end='2024-12-31', freq='D')
datos=pd.Series(range(len(fecha)),index=fecha)

print(datos)
```

```python
#Media de los datos agrupando por fechas
datos_mensuales = datos.resample('ME').mean()
display(datos_mensuales)
```

```python
#Ventanas moviles rolling windows
datos_rolling = datos.rolling(window=7).mean()
display(datos_rolling)
```
