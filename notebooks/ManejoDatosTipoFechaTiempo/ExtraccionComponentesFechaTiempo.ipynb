```python
import pandas as pd
```

```python
# Crear 10 fechas de ejemplo
df = pd.DataFrame({
    "fecha": ["2025/01/01", "2025/01/02", "2025/01/03",
              "2025/01/04", "2025/01/05", "2025/01/06",
              "2025/01/07", "2025/01/08", "2025/01/09", "2025/01/10"]
})
display(df)
```

```python
#Extraer componentes
# Extraer año, mes y día

# Convertir la columna a datetime
df['fecha'] = pd.to_datetime(df['fecha'], format="%Y/%m/%d", errors='coerce')
df['Año'] = df['fecha'].dt.year
df['Mes'] = df['fecha'].dt.month
df['Día'] = df['fecha'].dt.day

display(df)
```

```python
# Crear un rango de fechas con hora, minutos y segundos (10 datos cada hora)
fechas_con_hora = pd.date_range(start="2025-01-01 08:00:00", periods=10, freq="D")

# Crear DataFrame en formato año/mes/día hora:minuto:segundo
df_horas = pd.DataFrame({
    "fecha_hora": fechas_con_hora.strftime("%Y/%m/%d %H:%M:%S")
})

print(df_horas)
```

```python
# Convertir la columna a datetime
df_horas['fecha_hora'] = pd.to_datetime(df_horas['fecha_hora'], format="%Y/%m/%d %H:%M:%S")

df_horas['Hora']=df_horas['fecha_hora'].dt.hour
df_horas['Minuto']=df_horas['fecha_hora'].dt.minute
df_horas['Segundo']=df_horas['fecha_hora'].dt.second
display(df_horas)
```

```python
df_horas['dias_semana_numero']=df_horas['fecha_hora'].dt.weekday
df_horas['dias_semana_nombre']=df_horas['fecha_hora'].dt.day_name()
df_horas['trimestre']=df_horas['fecha_hora'].dt.quarter

display(df_horas)
```

```python
from babel.dates import format_datetime

#Pasar a español los dias de la semana
df_horas['Dia_español']=df_horas['fecha_hora'].apply(lambda x:format_datetime(x,"EEEE", locale="es"))
display(df_horas)
```

```python
df_horas['dia_año']=df_horas['fecha_hora'].dt.dayofyear
df_horas['Semana_Año']=df_horas['fecha_hora'].dt.isocalendar().week

display(df_horas)
```

```python
#Combinacion avanzados para hacer analisis de datos 
tendencias_dia_semana=df_horas.groupby('dias_semana_nombre').size()
print(tendencias_dia_semana)
```
