```python
import pandas as pd
```

```python
fechas = [
    "2025-09-01",              # ISO 8601
    "09/02/2025",              # Latinoamericano
    "09/03/2025",              # Estadounidense
    "09-04-25",                # Corto, año 2 dígitos
    "September 5, 2025",      # Inglés con nombre de mes   
    "20250906",                # Compacto
    "2025/09/07 14:30:00",     # Con hora (24h)
    "09-08-2025 02:30 PM",     # Con hora (12h AM/PM)
    "Fri, 09 Sep 2025",        # Abreviado con día semana  
   
]

df=pd.DataFrame({'fecha': fechas})
display(df)
```

```python
df['fecha']= pd.to_datetime(df['fecha'], format='mixed')
display(df)
```

```python
df['fecha_formateada']= df['fecha'].dt.strftime('%d-%m-%Y')
display(df)
```

```python
#Manejo de fechas invalidades o faltantes
fechas_invalidas = [
    "2025-09-01",              # ISO 8601
    "09/55/2025",              # Latinoamericano
    "09/03/2025",              # Estadounidense
    "09-04-25",                # Corto, año 2 dígitos
    "September 5, 2025",      # Inglés con nombre de mes   
    "20255566",                # Compacto
    "2025/09/07 14:30:00",     # Con hora (24h)
    "09-08-2025 02:30 PM",     # Con hora (12h AM/PM)
    "2024/99/99",        # Abreviado con día semana  
   
]
df=pd.DataFrame({'fecha': fechas_invalidas})
df['fecha']=pd.to_datetime(df['fecha'],errors='coerce')

display(df)
```

```python
#Datos con zonas horarios diferentes
df =pd.DataFrame({
    'fecha':pd.to_datetime(['2024-08-01 12:00:00','2025-08-02 12:00:00'])  
})
df=df.set_index('fecha') #Seteando como indece a la fecha
df=df.tz_localize('UTC').tz_convert('America/New_York')
display(df)
```
