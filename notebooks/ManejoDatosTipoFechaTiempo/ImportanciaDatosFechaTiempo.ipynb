```python
import pandas as pd
```

```python
fechas =pd.date_range(start='2023-01-01', periods=10,freq='D')
precios = [100,200,300,400,125,255,800,900,450,870]
serie_temporal = pd.Series(data=precios,index=fechas)
print(serie_temporal)

```

```python
primera_mitad=serie_temporal['2023-01-01':'2023-01-05']
segunda_mitad=serie_temporal['2023-1-06':'2023-01-10']
print('Primera mitad \n', primera_mitad)
print('Segunda mitad \n', segunda_mitad)
```

```python
#Detectar eventos y anomalias
#el método .diff() calcula la diferencia entre un valor y el valor anterior en una serie o columna numérica.
cambios = serie_temporal.diff().abs()
cambios_significativos=cambios[cambios>100]
print("cambios significativos:\n", cambios_significativos)
```

```python
#Agregaciones en los tiempos en los detalles
serie_semanal = serie_temporal.resample('W').sum()
print('Agregacion semanal\n', serie_semanal)
```
