```python
#Contar los que tiene M al inicio aplicar un filtro
filtro = serie_nombres.str.startswith('M')
serie_nombres[filtro]
```

```python
#Contar cuantos nombres empiezan con M
filtro = serie_nombres.str.startswith('M')
nombres_M = serie_nombres[filtro]
cantidad = filtro.sum()
print(f"Cantidad de nombres que empiezan con 'M': {cantidad}")  
```

```python
#Metodo contains() busca la letra en cualquier parte del texto de la serie
filtro = serie_nombres.str.contains('a')
nombres_con_M = serie_nombres[filtro]
print(f"Nombres que contienen la letra 'M':\n{nombres_con_M}")
```

```python
#Metodo que terminen con la letra a
filtro = serie_nombres.str.endswith('a')
nombres_con_a = serie_nombres[filtro]
print(f"Nombres que terminan con la letra 'a':\n{nombres_con_a}")
```

```python
#Filtros avanzado con regex
filtro = serie_nombres.str.contains('^M', regex=True)
nombres_M = serie_nombres[filtro]
print(f"Nombres que empiezan con 'M':\n{nombres_M}")

serie_nombres.str.contains('a$', regex=True)   # termina en a
print(f"Nombres que terminan con 'a':\n{nombres_con_a}")
filtro = serie_nombres.str.contains('M.*a', regex=True) # empieza con M y termina en a
nombres_con_M_y_a = serie_nombres[filtro]
print(f"Nombres que empiezan con 'M' y terminan con 'a':\n{nombres_con_M_y_a}")
filtro = serie_nombres.str.contains('^[A-Za-z]{4}$', regex=True) # exactamente 4 letras
nombres_4_letras = serie_nombres[filtro]
print(f"Nombres que tienen exactamente 4 letras:\n{nombres_4_letras}")
filtro = serie_nombres.str.contains('^[A-Za-z]+$', regex=True) # solo letras
nombres_solo_letras = serie_nombres[filtro]
print(f"Nombres que solo contienen letras:\n{nombres_solo_letras}")

```
